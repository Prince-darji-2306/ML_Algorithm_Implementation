# Support Vector Regressor (Primal Gradient Descent)
Simplified implementation using epsilon-insensitive loss with explanatory comments.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import r2_score
```

## 1. Mock Data Generation

```python
def get_data(n=100):
    X = np.sort(np.random.rand(n, 1) * 5, axis=0)
    y = np.sin(X).ravel() + np.random.normal(0, 0.1, n)
    return X, y

X, y = get_data()
plt.scatter(X, y, color='blue', alpha=0.5)
plt.title("Generated Regression Data")
plt.show()
```

## 2. Implementation

```python
class SVR:
    def __init__(self, lr=0.01, C=1.0, epsilon=0.1, iters=1000):
        self.lr = lr # Learning rate
        self.C = C # Cost parameter (regularization trade-off)
        self.epsilon = epsilon # Error threshold for insensitive loss
        self.iters = iters

    def fit(self, X, y):
        # Initialize parameters
        n, d = X.shape
        self.w = np.zeros(d)
        self.b = 0

        for _ in range(self.iters):
            for i, x in enumerate(X):
                prediction = np.dot(x, self.w) + self.b
                # If error is more than epsilon, we update
                if y[i] - prediction > self.epsilon:
                    self.w += self.lr * (self.C * x - self.w)
                    self.b += self.lr * self.C
                elif prediction - y[i] > self.epsilon:
                    self.w += self.lr * (-self.C * x - self.w)
                    self.b -= self.lr * self.C
                else:
                    # Errors within epsilon only face L2 shrinkage
                    self.w -= self.lr * self.w

    def predict(self, X):
        # Predicted target value
        return np.dot(X, self.w) + self.b
```

## 3. Training

```python
model = SVR(iters=2000)
model.fit(X, y)
preds = model.predict(X)
print(f"R2 Score: {r2_score(y, preds):.4f}")
```

## 4. Visualization

```python
plt.scatter(X, y, color='blue', alpha=0.5, label='Actual')
plt.plot(X, preds, color='red', lw=2, label='Predicted')
plt.title("SVR Regression Line")
plt.legend()
plt.show()
```
