# Support Vector Classifier (Primal Gradient Descent)
Simplified implementation using hinge loss with explanatory comments.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Mock Data Generation

```python
def get_data(n=100):
    X = np.random.randn(n, 2)
    y = np.where(X[:, 0] + X[:, 1] > 0, 1, -1)
    return X, y

X, y = get_data()
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm')
plt.title("Generated Data")
plt.show()
```

## 2. Implementation

```python
class SVMCLF:
    def __init__(self, lr=0.01, lambd=0.01, iters=1000):
        self.lr = lr # Learning rate
        self.lambd = lambd # Regularization strength
        self.iters = iters

    def fit(self, X, y):
        # Initialize parameters
        n, d = X.shape
        self.w = np.zeros(d)
        self.b = 0

        for _ in range(self.iters):
            for i, x in enumerate(X):
                # Hinge loss condition: target * prediction >= 1
                if y[i] * (np.dot(x, self.w) - self.b) >= 1:
                    # Inside safe margin: only apply L2 weight decay
                    self.w -= self.lr * (2 * self.lambd * self.w)
                else:
                    # Outside margin: update weights/bias using gradient
                    self.w -= self.lr * (2 * self.lambd * self.w - np.dot(x, y[i]))
                    self.b -= self.lr * y[i]

    def predict(self, X):
        # Simple prediction based on decision line
        return np.sign(np.dot(X, self.w) - self.b)
```

## 3. Training

```python
model = SVMCLF()
model.fit(X, y)
preds = model.predict(X)
acc = np.mean(preds == y)
print(f"Accuracy: {acc * 100:.2f}%")
```

## 4. Visualization

```python
def plot_boundary(X, y, model):
    h = .02
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, h), np.arange(y_min, y_max, h))
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)
    plt.contourf(xx, yy, Z, cmap='coolwarm', alpha=0.3)
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', edgecolors='k')
    plt.title("SVM Decision Boundary")
    plt.show()

plot_boundary(X, y, model)
```
